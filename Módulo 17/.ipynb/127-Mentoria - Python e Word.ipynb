## 📄 1. Criando um Documento e um Parágrafo Básico

Com a biblioteca `python-docx`, a gente monta o documento inteiro dentro do Python (ele só vira um arquivo `.docx` de verdade quando manda salvar) — dá pra gerar cartas, relatórios e contratos automaticamente, sem abrir o Word.

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `Document()` | cria um documento Word em branco, ainda só "na memória" do Python |
| `documento.add_paragraph(texto)` | adiciona um parágrafo de texto ao documento e devolve o objeto `Paragraph` criado |
| `documento.save(caminho)` | grava o documento na pasta, gerando o arquivo `.docx` de verdade |

> 💡 **Dica:** dá pra usar f-string dentro do texto do parágrafo, igual em qualquer outra string — inclusive pra jogar variáveis (tipo um faturamento calculado) direto na carta.

````python
from docx import Document
from cores import *

# criando o documento no Python
documento = Document()

faturamento = 1000

# aqui você edita tudo o que você quer
texto = f"""\nPrezado(a) Lira,

Informamos que o faturamento da empresa referente ao dia de ontem foi de:
R${faturamento}.

Atenciosamente, Lucas Paguetti"""

paragrafo = documento.add_paragraph(texto.title()) 
print(texto.title())
````

## 🎨 2. Formatação de Texto com Estilos Personalizados

Dá pra criar um **estilo próprio** (fonte, tamanho, cor, negrito etc.) e aplicar num parágrafo, ou simplesmente usar um dos estilos prontos que já vêm no Word (tipo os `Heading`).

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `documento.styles.add_style(nome, WD_STYLE_TYPE.PARAGRAPH)` | cria um novo estilo de parágrafo com nome próprio |
| `estilo.font.name` / `.size` / `.bold` / `.italic` / `.underline` | define fonte, tamanho (`Pt`), negrito, itálico e sublinhado do estilo |
| `estilo.font.color.rgb` | define a cor da fonte, em `RGBColor(r, g, b)` |
| `paragrafo.style = estilo` | aplica o estilo criado ao parágrafo |
| `documento.add_paragraph(texto, "Heading 5")` | usa direto um estilo pronto do Word, sem precisar criar um novo |

> ⚠️ **Atenção:** `Pt`, `RGBColor` e `Cm` vêm de `docx.shared`, e `WD_STYLE_TYPE` vem de `docx.enum.style` — são importações separadas da classe `Document`.

````python
from docx.shared import Pt, RGBColor, Cm # valores de formatação
from docx.enum.style import WD_STYLE_TYPE # criar seus próprios estilos

paragrafo.style = documento.styles.add_style("EstiloLira", WD_STYLE_TYPE.PARAGRAPH)
paragrafo.style.font.name = "Algerian"
paragrafo.style.font.size = Pt(15)
paragrafo.style.font.bold = True
paragrafo.style.font.italic = True
paragrafo.style.font.underline = True
paragrafo.style.font.color.rgb = RGBColor(255, 0, 0)
````

````python
paragrafo = documento.add_paragraph("PS: A quantidade de produtos vendidos foi de 10", "Heading 5")
````

````python
# implementando tudo que fizemos no Python no Texto.docx
documento.save("../.spec/Texto.docx")
````

## ✏️ 4. Editando Trechos Dentro do Mesmo Parágrafo (add_run)

Quando só uma parte da frase precisa de uma formatação diferente (ex.: destacar só o valor do faturamento em negrito), não precisa quebrar o texto em vários parágrafos — dá pra usar `add_run` pra criar um "trecho" formatado dentro do mesmo parágrafo.

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `paragrafo.add_run(texto)` | adiciona um novo trecho (`Run`) de texto dentro do mesmo parágrafo, e devolve esse trecho |
| `run.bold = True` | formata só o trecho criado pelo `add_run`, sem afetar o resto do parágrafo |

> 💡 **Dica:** dá pra encadear vários `add_run` seguidos no mesmo parágrafo, cada um com a sua própria formatação.

````python
from docx import Document

# criando o documento no Python
documento = Document()

faturamento = 1000

# aqui você edita tudo o que você quer
texto = f"""Fala Lira,

O faturamento da empresa ontem foi de """

texto_final = "Tamo junto, abs."


paragrafo = documento.add_paragraph(texto) 
paragrafo.add_run(f"R${faturamento}").bold = True

paragrafo = documento.add_paragraph(texto_final)

documento.save("../.spec/Texto.docx")
````

## 📐 5. Controle de Margem e Seções

Todo documento novo já nasce com uma seção (`section`) padrão — é nela que configuramos as margens da página.

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `documento.sections` | lista as seções do documento |
| `secao.top_margin` / `.bottom_margin` / `.left_margin` / `.right_margin` | definem as margens da página, em `Cm(valor)` (importado de `docx.shared`) |
| `paragrafo.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER` | alinha o parágrafo (importado de `docx.enum.text`) |

> 💡 **Dica:** como um documento pode ter mais de uma seção (ex.: quebra de seção pra mudar orientação da página no meio do arquivo), o `for` percorre todas de uma vez.

````python
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT

# criando o documento no Python
documento = Document()

for secao in documento.sections:
    secao.top_margin = Cm(0.5)
    secao.bottom_margin = Cm(1)
    secao.left_margin = Cm(1)
    secao.right_margin = Cm(1)


faturamento = 1000

# aqui você edita tudo o que você quer
texto = f"""Fala Lira,

O faturamento da empresa ontem foi de R${faturamento}

Tamo junto, abs.
"""

paragrafo = documento.add_paragraph(texto) 
paragrafo.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER

documento.save("../.spec/Texto.docx")
````

## 🖼️ 6. Inserindo Imagens

## 📊 7. Inserindo Tabelas

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `documento.add_table(rows=1, cols=3, style="Nome do Estilo")` | cria uma tabela com número de linhas/colunas definido, já usando um estilo pronto do Word |
| `table.rows[0].cells` | acessa as células da primeira linha (geralmente o cabeçalho) |
| `table.add_row().cells` | adiciona uma nova linha à tabela e devolve as células dela pra preencher |

> 💡 **Dica:** o `style="Light Grid Accent 1"` é um dos estilos de tabela que já vêm prontos no Word — dá pra ver todos com o `documento.styles` (próxima seção).

````python
documento = Document()

records = (
    (3, '101', 'Spam'),
    (7, '422', 'Eggs'),
    (4, '631', 'Spam, spam, eggs, and spam')
)

table = documento.add_table(rows=1, cols=3, style="Light Grid Accent 1")
hdr_cells = table.rows[0].cells
hdr_cells[0].text = 'Qty'
hdr_cells[1].text = 'Id'
hdr_cells[2].text = 'Desc'
for qty, id, desc in records:
    row_cells = table.add_row().cells
    row_cells[0].text = str(qty)
    row_cells[1].text = id
    row_cells[2].text = desc
    
    
documento.save("../.spec/Texto.docx")
````

## 🔄 9. Adaptando um Documento Existente (Substituição de Placeholders)

Uma forma de gerar vários documentos parecidos (contratos, declarações) é escrever um modelo com "códigos" no lugar dos dados que mudam (ex.: `XXXX`, `YYYY`) e trocar esses códigos pelos valores de verdade via Python.

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `Document("Contrato.docx")` | abre um contrato/modelo já pronto, escrito com placeholders no lugar dos dados variáveis |
| `contrato.paragraphs` | lista todos os parágrafos do documento, na ordem em que aparecem |
| `paragrafo.text.replace(codigo, valor)` | troca o texto do placeholder pelo valor de verdade, parágrafo por parágrafo |

> ⚠️ **Atenção:** esse `replace()` funciona no texto do parágrafo inteiro — se o placeholder estiver "quebrado" em `runs` diferentes (ex.: metade em negrito, metade não), a substituição pode não encontrar o texto completo.

````python
from datetime import datetime

contrato = Document("../.spec/Contrato.docx")

nome = "Lira da Hashtag"
item1 = "Serviço de Treinamento em Excel"
item2 = "Apostila Completa de Excel"
item3 = "Serviço de Treinamentos de Python"

dicionario_valores = {
    "XXXX": nome,
    "YYYY": item1,
    "ZZZZ": item2,
    "WWWW": item3,
    "DD": str(datetime.now().day),
    "MM": str(datetime.now().month),
    "AAAA": str(datetime.now().year),
}

# tem alguma função replace? -> é executada por parágrafo
# percorrer as linhas
    # se a linha tiver o texto xxxx substitui pelo nome
for paragrafo in contrato.paragraphs:
    # para cada placeholder do dicionario
    for codigo in dicionario_valores:
        if codigo in paragrafo.text:
            paragrafo.text = paragrafo.text.replace(codigo, dicionario_valores[codigo])

contrato.save(f"../.spec/Contrato Atualizado - {nome}.docx")
````
