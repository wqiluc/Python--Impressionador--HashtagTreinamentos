## 📖 1. Lendo o Texto de um PDF

`pdfplumber` é a biblioteca certa quando o objetivo é **ler** o conteúdo de um PDF (texto, tabelas, posição de cada palavra na página) — bem diferente do `pypdf`, que serve pra **manipular** o arquivo em si (juntar, separar, proteger).

| Método 🔑 | O que faz 🔓 |
|---|---|
| `pdfplumber.open(caminho)` | abre o PDF e devolve um objeto navegável por página |
| `pdf.pages` | lista com todas as páginas do documento, cada uma um objeto `Page` |
| `pagina.extract_text()` | devolve todo o texto da página como uma única string |

> 💡 **Dica:** use `with pdfplumber.open(...) as pdf:` — ele fecha o arquivo sozinho no final, igual `open()` normal.


````python
import pdfplumber
from cores import *

with pdfplumber.open("../.spec/DANFEBrota.pdf") as pdf:
    primeira_pagina = pdf.pages[0]
    texto = primeira_pagina.extract_text()

print(f"{CinzaClaro}{Reset}{MagentaClaro}{texto[0:].title()}{Reset}")
````

## 🔢 2. Navegando pelas Páginas do Documento

Um relatório de resultados tem dezenas de páginas — dá pra percorrer todas elas com um `for`, igual uma lista comum.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `len(pdf.pages)` | quantidade total de páginas do PDF |
| `for pagina in pdf.pages:` | percorre as páginas na ordem em que aparecem no documento |
| `pagina.page_number` | número da página (começando em 1, diferente do índice da lista que começa em 0) |

> ⚠️ **Atenção:** `pdf.pages[0]` é a primeira página, mas `pagina.page_number` dessa mesma página já vem como `1`.


````python
caminho_3t20 = "../.spec/Integração Python e PDF - Como funciona - MGLU_ER_3T20_POR.pdf"

with pdfplumber.open(f"{caminho_3t20}") as pdf:
    print(f"{CinzaClaro}Total de páginas:{Reset} {MagentaClaro}{len(pdf.pages)}{Reset}\n")
    for indice_pagina, (pagina) in enumerate(pdf.pages[:5]):
        texto = pagina.extract_text() or ""
        primeira_linha = texto.split("\n")[0]
        print(f"{CinzaClaro}Página {pagina.page_number}:{Reset} {MagentaClaro}{primeira_linha}{Reset}")
````

## 📊 3. Extraindo Tabelas de Dentro do PDF

Relatórios financeiros costumam trazer os números em tabelas — `extract_tables()` reconhece as linhas/colunas da página e devolve cada tabela como uma lista de listas, pronta pra virar um `DataFrame`.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `pagina.extract_tables()` | devolve **todas** as tabelas encontradas na página, cada uma como lista de linhas |
| `tabela[linha][coluna]` | acessa uma célula específica da tabela extraída |

> 💡 **Dica:** uma página pode ter mais de uma tabela — por isso `extract_tables()` sempre devolve uma lista de tabelas, mesmo quando só existe uma.


````python
with pdfplumber.open(caminho_3t20) as pdf:
    pagina_destaques = pdf.pages[4]
    tabela_destaques = pagina_destaques.extract_tables()[0]

for indice_linha, linha in enumerate(tabela_destaques[:8]):
    print(f"{CinzaClaro}{Reset}{MagentaClaro}{linha}{Reset}")
````

## 🔍 4. Localizando a Página Certa por Palavra-chave

O problema de "chutar" o índice da página (como fizemos acima com `pdf.pages[4]`) é que ele muda de arquivo pra arquivo — no relatório do 3T20 a tabela de destaques está na página 5 (índice `4`), mas no do 4T20 ela está na página 7 (índice `6`). A solução é **procurar** a página certa pelo texto, em vez de fixar o número.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `palavra in texto` | testa se um trecho de texto está contido na página (funciona igual em qualquer string) |
| `enumerate(pdf.pages)` | devolve o índice junto com cada página, útil pra guardar "achei na página X" |

> ⚠️ **Atenção:** `pagina.extract_text()` devolve `None` em páginas sem texto (ex.: uma página só com imagem) — por isso usar `pagina.extract_text() or ""` evita erro na hora de testar `palavra in texto`.


````python
def localizar_pagina_destaques(caminho_pdf, palavra_chave="Vendas Totais"):
    with pdfplumber.open(caminho_pdf) as pdf:
        for indice, pagina in enumerate(pdf.pages):
            texto = pagina.extract_text() or ""
            if (palavra_chave in texto):
                return indice
    return None

caminho_4t20 = r"../.spec/Integração Python e PDF - Como funciona - MGLU_ER_4T20_POR.pdf"

print(f"{CinzaClaro}Página com destaques no 3T20:{Reset} {MagentaClaro}{localizar_pagina_destaques(caminho_3t20)}{Reset}")
print(f"{CinzaClaro}Página com destaques no 4T20:{Reset} {MagentaClaro}{localizar_pagina_destaques(caminho_4t20)}{Reset}")
````

## 🎯 5. Mentoria: Consolidando Dados de Vários PDFs num CSV

Nessa mentoria, juntamos tudo que vimos até aqui pra ler os **dois** relatórios trimestrais (3T20 e 4T20), sem abrir arquivo por arquivo manualmente, e gerar uma única planilha comparando os trimestres — igual fizemos com as notas fiscais em XML.

**A lógica:**
- Pra cada PDF, localizar a página de destaques procurando por `"Vendas Totais"` (item 4).
- Extrair a tabela dessa página (item 3) e pegar só as linhas que interessam num dicionário.
- Juntar os dicionários de todos os PDFs numa lista e transformar em `DataFrame` do pandas.
- Salvar o resultado em `.csv`, pronto pra abrir no Excel ou reler depois com `pd.read_csv`.


````python
import pandas as pd

def extrair_destaques(caminho_pdf):
    indice_pagina = localizar_pagina_destaques(caminho_pdf)

    with pdfplumber.open(caminho_pdf) as pdf:
        tabela = pdf.pages[indice_pagina].extract_tables()[0]

    # linha[1] = nome do indicador, linha[2] = valor do trimestre (colunas da tabela)
    indicadores = {linha[1].strip(): linha[2] for linha in tabela if linha[1] and linha[2]}

    return {
        "trimestre": tabela[0][2],
        "receita_liquida": indicadores.get("Receita Líquida"),
        "lucro_bruto": indicadores.get("Lucro Bruto"),
        "ebitda": indicadores.get("EBITDA"),
        "lucro_liquido": indicadores.get("Lucro Líquido"),
    }

arquivos = [caminho_3t20, caminho_4t20]
destaques = [extrair_destaques(arquivo) for indice_arquivo, arquivo in enumerate(arquivos)]

df_destaques = pd.DataFrame(destaques)
df_destaques.to_csv("../.spec/Resultado_Divulgacao_Resultados.csv", index=False, sep=";", encoding="latin-1")
````

````python
# lendo de volta o csv que acabamos de gerar
caminho_csv = "../.spec/Resultado_Divulgacao_Resultados.csv"
pd.read_csv(caminho_csv, sep=";", encoding="latin-1")

# tratamento dos nomes das colunas, para fins intuitivos
df_destaques_updated = df_destaques.rename(columns=
                                            {
                                                "trimestre": "Trimestre",
                                                "receita_liquida": "Receita Líquida",
                                                "lucro_bruto": "Lucro Bruto",
                                                "ebitda": "Ebtida",
                                                "lucro_liquido": "Lucro Líquido" 
                                            }
                                           )

display(df_destaques_updated)
````

## 📎 6. Juntando Vários PDFs em Um Só

Pra manipular o arquivo PDF em si (não só ler o conteúdo), a biblioteca é a `pypdf`. Ela trabalha com dois objetos principais: `PdfReader` (lê um PDF existente) e `PdfWriter` (monta um PDF novo, página por página).

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `PdfReader(caminho)` | abre um PDF existente pra leitura |
| `PdfWriter()` | cria um PDF novo, vazio, pronto pra receber páginas |
| `writer.append(caminho)` | adiciona **todas** as páginas de um PDF ao final do `writer` |
| `writer.write(caminho)` | grava o PDF montado no `writer` em um arquivo `.pdf` de verdade |

> 💡 **Dica:** `writer.append()` aceita tanto um caminho de arquivo quanto um objeto `PdfReader` já aberto.


````python
from pypdf import PdfReader, PdfWriter

writer = PdfWriter()
writer.append("../.spec/DANFEBrota.pdf")
writer.append("../.spec/DANFENespresso.pdf")

with open("../.spec/NotasFiscais_Unificado.pdf", "wb") as arquivo_saida:
    writer.write(arquivo_saida)

print(f"{CinzaClaro}Páginas no PDF unificado:{Reset} {MagentaClaro}{len(PdfReader('../.spec/NotasFiscais_Unificado.pdf').pages)}{Reset}")
````

## ✂️ 7. Separador de Páginas de um PDF

Dá pra fazer o caminho inverso da junção: em vez de um `PdfWriter` só, monta-se **um `PdfWriter` por página**, cada um salvo no seu próprio arquivo — útil pra distribuir só a página que interessa de um PDF grande, sem mandar o documento inteiro.

| Classe/Método 🔑 | O que faz 🔓 |
|---|---|
| `enumerate(reader.pages)` | percorre as páginas do PDF já trazendo o índice de cada uma |
| `writer.add_page(pagina)` | adiciona **uma única** página (em vez do PDF inteiro) a um `PdfWriter` novo, exclusivo dela |
| `os.makedirs(pasta, exist_ok=True)` | garante que a pasta de saída existe antes de salvar os arquivos, sem dar erro se ela já existir |

> 💡 **Dica:** `str(indice + 1).zfill(2)` nomeia os arquivos com zero à esquerda (`pagina_01.pdf`, `pagina_02.pdf`...), o que mantém a ordem certa ao listar a pasta.


````python
import os
from pypdf import PdfReader, PdfWriter

def separar_paginas(caminho_pdf, pasta_saida):
    os.makedirs(pasta_saida, exist_ok=True)
    reader = PdfReader(caminho_pdf)

    for indice, pagina in (enumerate(reader.pages)):
        writer = PdfWriter()
        writer.add_page(pagina)
        nome_arquivo = f"{pasta_saida}/pagina_{str(indice + 1).zfill(2)}.pdf"
        with open(nome_arquivo, "wb") as arquivo_saida:
            writer.write(arquivo_saida)

# separando de volta o PDF que acabamos de unificar no item anterior
pasta_paginas = "../.spec/NotasFiscais_Paginas"
separar_paginas("../.spec/NotasFiscais_Unificado.pdf", pasta_paginas)

for indice_arquivos, arquivos in (enumerate(pasta_paginas)):
    print(f"{CinzaClaro}Arquivos gerados:{Reset} {MagentaClaro}{sorted(os.listdir(pasta_paginas))[0:]}{Reset}")
    break
````

## 🔒 8. Protegendo um PDF com Senha

Pra fechar, dá pra exigir uma senha antes de abrir o PDF — útil pra relatórios com informação sensível (dados de cliente, contrato, folha de pagamento) que vão ser enviados por e-mail. A senha é aplicada no `writer` **antes** de salvar, então dá pra ir direto da página extraída pro arquivo já protegido, sem deixar uma cópia sem senha salva no meio do caminho.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `writer.encrypt(senha)` | define uma senha de abertura pro PDF que está sendo montado no `writer`, antes mesmo de salvar |
| `reader.is_encrypted` | indica se o PDF aberto está protegido por senha |
| `reader.decrypt(senha)` | destrava um `PdfReader` protegido, liberando leitura/extração normalmente |

> ⚠️ **Atenção:** sem chamar `reader.decrypt(senha)` primeiro, tentar ler página ou texto de um PDF protegido lança erro.


````python
indice_pagina_3t20 = localizar_pagina_destaques(caminho_3t20)

reader = PdfReader(caminho_3t20)

writer = PdfWriter()
writer.add_page(reader.pages[indice_pagina_3t20])
writer.encrypt("hashtag123")

with open("../.spec/Destaques_3T20_Protegido.pdf", "wb") as arquivo_saida:
    writer.write(arquivo_saida)

reader_protegido = PdfReader("../.spec/Destaques_3T20_Protegido.pdf")
print(f"{CinzaClaro}Está protegido?{Reset} {VerdeClaro}{reader_protegido.is_encrypted}{Reset}")

reader_protegido.decrypt("hashtag123")
print(f"{CinzaClaro}{Reset}{MagentaClaro}{reader_protegido.pages[0].extract_text()[:120]}{Reset}")
````
