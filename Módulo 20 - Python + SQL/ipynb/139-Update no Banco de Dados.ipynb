> O "U" do CRUD: alterar linhas que já existem, sempre com `UPDATE ... SET ... WHERE`. O `WHERE` aqui é ainda mais crítico que no `SELECT` — sem ele, o `UPDATE` altera **a tabela inteira**.

## 🔌 1. Conectando

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
cursor = conexao.cursor()
```

## 🎯 2. UPDATE de um Único Registro

Filtrando por `Id` (a chave primária), o `UPDATE` atinge exatamente uma linha.

| Cláusula 🔑 | Papel 🔓 |
|---|---|
| `SET coluna = ?` | qual valor muda |
| `WHERE Id = ?` | qual linha muda |

```python
cursor.execute("SELECT Id, Produto, ValorUnitario FROM dbo.Vendas WHERE Produto = ?", "Mouse sem Fio")
antes = cursor.fetchone()
print(f"{CinzaClaro}Antes:{Reset} {VerdeClaro}{antes.Produto}{Reset} — R${antes.ValorUnitario}")

cursor.execute(
    "UPDATE dbo.Vendas SET ValorUnitario = ? WHERE Id = ?",
    79.90, antes.Id
)
conexao.commit()

cursor.execute("SELECT Produto, ValorUnitario FROM dbo.Vendas WHERE Id = ?", antes.Id)
depois = cursor.fetchone()
print(f"{CinzaClaro}Depois:{Reset} {VerdeClaro}{depois.Produto}{Reset} — R${depois.ValorUnitario}")
```

## 📦 3. UPDATE em Massa com WHERE

O `WHERE` não precisa filtrar uma linha só — qualquer condição que bata com várias linhas faz o `UPDATE` valer pra todas elas de uma vez.

```python
cursor.execute("SELECT COUNT(*) AS Total FROM dbo.Vendas WHERE Vendedor = ?", "Ana Souza")
total_antes = cursor.fetchone().Total

cursor.execute(
    "UPDATE dbo.Vendas SET Vendedor = ? WHERE Vendedor = ?",
    "Ana Souza Ferreira", "Ana Souza"
)
conexao.commit()

print(f"{VerdeClaro}{cursor.rowcount} linhas atualizadas{Reset} {CinzaClaro}(vendedora renomeada nas {total_antes} vendas dela).{Reset}")
```

## ⚠️ 4. O Perigo do UPDATE sem WHERE

Sem `WHERE`, o `SET` vale pra **toda linha da tabela**. Vale rodar sempre um `SELECT` com a mesma condição antes, pra conferir quantas linhas serão afetadas.

> 💡 **Regra prática:** antes de qualquer `UPDATE`/`DELETE` com `WHERE`, rode um `SELECT COUNT(*) ... WHERE` com a mesma condição. Se o número bater com o esperado, aí sim execute o `UPDATE`/`DELETE`.

```python
cursor.execute("SELECT COUNT(*) AS Total FROM dbo.Vendas WHERE Categoria = ?", "Eletrônicos")
quantas_linhas_seriam_afetadas = cursor.fetchone().Total

print(f"{AmareloClaro}Um UPDATE com essa condição afetaria {quantas_linhas_seriam_afetadas} linhas{Reset} {CinzaClaro}— confira sempre antes de rodar de verdade.{Reset}")

conexao.close()
```

Update feito, com o cuidado de sempre conferir o `WHERE` antes. Falta só o "D" do CRUD.
