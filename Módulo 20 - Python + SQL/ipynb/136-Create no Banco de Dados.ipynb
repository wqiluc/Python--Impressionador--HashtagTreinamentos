## 🔌 1. Conectando

Mesma conexão de sempre, via `conexao.py`.

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
cursor = conexao.cursor()
```

## 🏗️ 2. CREATE TABLE

A tabela `Vendas` guarda cada venda: produto, categoria, quantidade, valor unitário, data e vendedor. `IDENTITY(1,1)` faz o `Id` se auto-incrementar — não precisamos informá-lo nos `INSERT`s.

| Coluna 🔑 | Tipo 🔓 | Papel 🔓 |
|---|---|---|
| `Id` | `INT IDENTITY` | Chave primária, gerada sozinha |
| `Produto` | `VARCHAR(100)` | Nome do produto vendido |
| `Categoria` | `VARCHAR(50)` | Categoria do produto |
| `Quantidade` | `INT` | Unidades vendidas |
| `ValorUnitario` | `DECIMAL(10,2)` | Preço de cada unidade |
| `DataVenda` | `DATE` | Data em que a venda aconteceu |
| `Vendedor` | `VARCHAR(100)` | Quem fez a venda |

```python
cursor.execute("""
IF OBJECT_ID('dbo.Vendas', 'U') IS NOT NULL
    DROP TABLE dbo.Vendas;

CREATE TABLE dbo.Vendas (
    Id INT IDENTITY(1,1) PRIMARY KEY,
    Produto VARCHAR(100) NOT NULL,
    Categoria VARCHAR(50) NOT NULL,
    Quantidade INT NOT NULL,
    ValorUnitario DECIMAL(10,2) NOT NULL,
    DataVenda DATE NOT NULL,
    Vendedor VARCHAR(100) NOT NULL
);
""")
conexao.commit()

print(f"{VerdeClaro}Tabela Vendas criada.{Reset}")
```

## ➕ 3. INSERT — Uma Linha por Vez

O jeito mais direto: um `INSERT INTO` com um `?` por coluna, e os valores na mesma ordem como argumentos de `execute()`.

```python
cursor.execute(
    """INSERT INTO dbo.Vendas 
    (Produto, Categoria, Quantidade, ValorUnitario, DataVenda, Vendedor)
    VALUES (?, ?, ?, ?, ?, ?)""",
    "Notebook Gamer", "Eletrônicos", 2, 4500.00, "2026-01-05", "Ana Souza"
)
conexao.commit()

print(f"{CinzaClaro}1 linha{Reset} {VerdeClaro}inserida.{Reset}")
```

## 📚 4. INSERT — Várias Linhas com executemany

Quando há várias linhas pra inserir de uma vez, `executemany()` roda o mesmo `INSERT` pra cada tupla de uma lista — muito mais rápido do que chamar `execute()` em loop.

| Método 🔑 | Quando usar 🔓 |
|---|---|
| `cursor.execute(sql, valores)` | Uma linha |
| `cursor.executemany(sql, lista_de_tuplas)` | Várias linhas de uma vez |

```python
novas_vendas = [
    ("Mouse sem Fio", "Eletrônicos", 5, 89.90, "2026-01-06", "Bruno Lima"),
    ("Cadeira Gamer", "Móveis", 1, 1299.00, "2026-01-06", "Ana Souza"),
    ("Monitor 27\"", "Eletrônicos", 3, 1199.00, "2026-01-07", "Carla Dias"),
    ("Teclado Mecânico", "Eletrônicos", 4, 349.90, "2026-01-08", "Bruno Lima"),
    ("Mesa de Escritório", "Móveis", 2, 749.00, "2026-01-08", "Carla Dias"),
    ("Headset", "Eletrônicos", 6, 259.90, "2026-01-09", "Ana Souza"),
    ("Luminária de Mesa", "Móveis", 3, 129.90, "2026-01-09", "Bruno Lima"),
]

cursor.executemany(
    """INSERT INTO dbo.Vendas 
    (Produto, Categoria, Quantidade, ValorUnitario, DataVenda, Vendedor)
    VALUES (?, ?, ?, ?, ?, ?)""", novas_vendas
)
conexao.commit()

print(f"{CinzaClaro}{len(novas_vendas)} linhas{Reset} {VerdeClaro}inseridas{Reset} {CinzaClaro}via executemany.{Reset}")
```

## ✅ 5. Conferindo o Resultado

Um `SELECT COUNT(*)` rápido confirma quantas linhas a tabela `Vendas` tem agora.

```python
cursor.execute("SELECT COUNT(*) AS Total FROM dbo.Vendas")
total = cursor.fetchone().Total

print(f"{CinzaClaro}Total de vendas cadastradas:{Reset} {MagentaClaro}{total}{Reset}")

conexao.close()
```
