## 🔌 1. Conectando

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL", autocommit=True)
cursor = conexao.cursor()
```

## 🛠️ 2. Construindo remover_insumo()

A função confere se o insumo existe antes de apagar — assim dá pra avisar quando alguém tenta remover algo que já não está no estoque, em vez de simplesmente não fazer nada silenciosamente.

| Retorno 🔑 | Situação 🔓 |
|---|---|
| `True` | insumo existia e foi removido |
| `False` | insumo não foi encontrado |

```python
def remover_insumo(nome):
    cursor.execute("SELECT Id FROM dbo.Estoque WHERE Insumo = ?", nome)
    insumo_existente = cursor.fetchone()

    if not insumo_existente:
        print(f"{VermelhoClaro}'{nome}' não está cadastrado no estoque.{Reset}")
        return False

    cursor.execute("DELETE FROM dbo.Estoque WHERE Id = ?", insumo_existente.Id)
    print(f"{VermelhoClaro}'{nome}' removido do estoque.{Reset}")
    return True
```

## ✅ 3. Testando com um Insumo que Existe

```python
remover_insumo("Fermento Químico")

print(f"\n{CinzaClaro}Estoque atual:{Reset}")
cursor.execute("SELECT Insumo, Quantidade FROM dbo.Estoque ORDER BY Insumo")
for linha in cursor.fetchall():
    print(f"  {VerdeClaro}{linha.Insumo}{Reset} — {linha.Quantidade}")
```

## ❓ 4. Testando com um Insumo que Não Existe

Chamando de novo pro mesmo insumo (já removido) — a função avisa em vez de tentar apagar algo que não existe.

```python
resultado = remover_insumo("Fermento Químico")
print(f"{CinzaClaro}Função retornou:{Reset} {MagentaClaro}{resultado}{Reset}")

conexao.close()
```
