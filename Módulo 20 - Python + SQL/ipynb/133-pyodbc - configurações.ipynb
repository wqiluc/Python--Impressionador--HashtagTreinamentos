## 🧩 1. As Duas Instalações Necessárias

`pyodbc` sozinho não conecta em nada — ele é só a ponte. Quem realmente sabe falar o protocolo do SQL Server é o **driver ODBC**, instalado no sistema operacional (não com `pip`).

| Peça 🔑 | O que é 🔓 | Como instala 🔓 |
|---|---|---|
| `pyodbc` | Biblioteca Python que usa a interface ODBC | `pip install pyodbc` |
| Driver ODBC do SQL Server | Componente do sistema que sabe o protocolo do SQL Server | Windows: instalador da Microsoft · Mac/Linux: `unixodbc` + `msodbcsql` |

> ⚠️ **Atenção:** em Mac e Linux, sem o driver ODBC instalado no sistema, `pyodbc` importa normalmente mas **toda tentativa de conexão falha**. É o erro mais comum de quem começa com `pyodbc` fora do Windows.

## 📦 2. Instalando o pyodbc

Na própria instalação, sem segredo:

```bash
pip install pyodbc
```

No Mac, o driver ODBC do SQL Server vem via Homebrew (a Microsoft publica um "tap" próprio):

```bash
brew tap microsoft/mssql-release https://github.com/Microsoft/homebrew-mssql-release
HOMEBREW_ACCEPT_EULA=Y ACCEPT_EULA=Y brew install unixodbc msodbcsql18
```

No Windows, basta baixar e rodar o instalador **"ODBC Driver for SQL Server"** direto no [site da Microsoft](https://learn.microsoft.com/sql/connect/odbc/download-odbc-driver-for-sql-server).

````python
import pyodbc
from cores import *

print(f"{CinzaClaro}Versão do pyodbc:{Reset} {VerdeClaro}{pyodbc.version}{Reset}")
````

## 🔍 3. Verificando os Drivers ODBC Instalados

Antes de tentar conectar, vale confirmar que o sistema enxerga o driver — `pyodbc.drivers()` lista tudo o que está registrado no ODBC do sistema operacional.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `pyodbc.drivers()` | devolve a lista de drivers ODBC instalados no sistema |

````python
drivers_instalados = pyodbc.drivers()

print(f"{CinzaClaro}Drivers ODBC encontrados:{Reset}")
for driver in drivers_instalados:
    print(f"{VerdeClaro}  - {driver}{Reset}")
````

## 🧵 4. Montando a String de Conexão

`pyodbc.connect()` recebe uma única string, com pares `chave=valor;` separados por `;`. Os principais parâmetros:

| Parâmetro 🔑 | Para que serve 🔓 |
|---|---|
| `DRIVER` | Nome exato do driver, entre chaves — igual aparece em `pyodbc.drivers()` |
| `SERVER` | Endereço do servidor (`host,porta` ou `host\\instância`) |
| `DATABASE` | Banco de dados dentro do servidor (opcional — sem isso, conecta no banco padrão) |
| `UID` / `PWD` | Usuário e senha |
| `TrustServerCertificate` | `yes` para aceitar certificado autoassinado (comum em ambiente local/Docker) |

````python
string_conexao = (
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=localhost,1433;"
    "UID=sa;"
    "PWD=HashtagSql#2026;"
    "TrustServerCertificate=yes;"
)

print(f"{CinzaClaro}String de conexão montada:{Reset}")
print(f"{MagentaClaro}{string_conexao}{Reset}")
````

## 🚀 5. Primeira Conexão de Verdade

Com a string pronta, `pyodbc.connect()` abre a conexão. Se não der erro, o servidor aceitou usuário e senha — é o sinal de que a configuração está correta.

````python
conexao = pyodbc.connect(string_conexao)
cursor = conexao.cursor()

cursor.execute("SELECT @@VERSION")
versao_servidor = cursor.fetchone()[0]

print(f"{CinzaClaro}Conectado com sucesso!{Reset}")
print(f"{VerdeClaro}{versao_servidor.splitlines()[0]}{Reset}")

conexao.close()
````

Deu certo — a conexão abriu, executou um `SELECT` e fechou sem erro. Esse é o esqueleto de tudo que vem a seguir. Só que repetir usuário/senha em toda célula não é uma boa prática — é exatamente o assunto do próximo notebook.
