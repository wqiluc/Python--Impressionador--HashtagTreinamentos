## 🧩 1. O Problema

Uma cozinha compra insumos (farinha, ovos, leite...), usa parte deles na produção do dia, e precisa saber a qualquer momento **quanto ainda tem de cada um**. Hoje isso é feito numa planilha, que trava com duas pessoas mexendo ao mesmo tempo e não avisa quando algo está acabando.

A proposta: um mini sistema em Python, com o SQL Server guardando o estado real do estoque, e funções Python simples pra cada operação do dia a dia.

## 🗄️ 2. A Tabela Estoque

| Coluna 🔑 | Tipo 🔓 | Papel 🔓 |
|---|---|---|
| `Id` | `INT IDENTITY` | Chave primária, gerada sozinha |
| `Insumo` | `VARCHAR(100)` | Nome do insumo (único — não pode repetir) |
| `Quantidade` | `DECIMAL(10,2)` | Quanto tem disponível, na unidade de medida |
| `UnidadeMedida` | `VARCHAR(20)` | `kg`, `L`, `un`, etc. |
| `Categoria` | `VARCHAR(50)` | Grupo do insumo (Laticínios, Grãos, Hortifruti...) |
| `AtualizadoEm` | `DATETIME` | Última vez que a linha mudou |

## 🛠️ 3. As Funções que Vamos Construir

| Notebook 🔑 | Função 🔓 | O que faz 🔓 |
|---|---|---|
| Conexão com o Banco | `conectar_estoque()` | Cria a tabela e popula o estoque inicial |
| Adicionando Insumo (CREATE) | `adicionar_insumo()` | Cadastra um insumo novo, ou soma quantidade se ele já existir |
| Deletar Insumo (DELETE) | `remover_insumo()` | Remove um insumo do controle |
| Consumir Insumo (UPDATE) | `consumir_insumo()` | Dá baixa numa quantidade usada na produção |
| Procurar Insumo (READ) | `buscar_insumo()` / `listar_estoque()` | Consulta o que tem, e alerta o que está acabando |

## 🧭 4. Como os Notebooks se Encaixam

Cada notebook usa a mesma tabela `Estoque`, criada no notebook de Conexão — o estado vai se acumulando de um notebook pro outro, igual aconteceu com `Vendas` nos notebooks de Create/Read/Update/Delete anteriores.
