## 🏗️ 1. CREATE TABLE Estoque

Igual ao padrão usado em `Vendas`: `IF OBJECT_ID(...) IS NOT NULL DROP TABLE` deixa a célula segura pra rodar de novo. `Insumo` como `UNIQUE` garante que não existam dois registros pro mesmo insumo — evitando estoque duplicado e "espalhado" em duas linhas.

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
cursor = conexao.cursor()

cursor.execute("""
IF OBJECT_ID('dbo.Estoque', 'U') IS NOT NULL
    DROP TABLE dbo.Estoque;

CREATE TABLE dbo.Estoque (
    Id INT IDENTITY(1,1) PRIMARY KEY,
    Insumo VARCHAR(100) NOT NULL UNIQUE,
    Quantidade DECIMAL(10,2) NOT NULL,
    UnidadeMedida VARCHAR(20) NOT NULL,
    Categoria VARCHAR(50) NOT NULL,
    AtualizadoEm DATETIME NOT NULL DEFAULT GETDATE()
);
""")
conexao.commit()

print(f"{VerdeClaro}Tabela Estoque criada.{Reset}")
```

## 🌱 2. Populando o Estoque Inicial

Um `executemany()` cadastra o ponto de partida do estoque — como se fosse a primeira contagem física feita na cozinha.

```python
estoque_inicial = [
    ("Farinha de Trigo", 50.0, "kg", "Grãos"),
    ("Ovos", 300.0, "un", "Laticínios"),
    ("Leite", 80.0, "L", "Laticínios"),
    ("Açúcar", 40.0, "kg", "Grãos"),
    ("Manteiga", 15.0, "kg", "Laticínios"),
    ("Fermento Biológico", 5.0, "kg", "Grãos"),
    ("Chocolate em Pó", 12.0, "kg", "Confeitaria"),
    ("Morango", 8.0, "kg", "Hortifruti"),
]

cursor.executemany(
    """INSERT INTO dbo.Estoque (Insumo, Quantidade, UnidadeMedida, Categoria)
       VALUES (?, ?, ?, ?)""",
    estoque_inicial
)
conexao.commit()

print(f"{VerdeClaro}{len(estoque_inicial)} insumos cadastrados no estoque inicial.{Reset}")
```

## ✅ 3. Conferindo o Resultado

```python
cursor.execute("SELECT Insumo, Quantidade, UnidadeMedida, Categoria FROM dbo.Estoque ORDER BY Categoria, Insumo")
for linha in cursor.fetchall():
    print(f"  {VerdeClaro}{linha.Insumo}{Reset} — {linha.Quantidade} {linha.UnidadeMedida} {CinzaEscuro}({linha.Categoria}){Reset}")

conexao.close()
```

Estoque de pé, com 8 insumos cadastrados. O próximo notebook constrói a função de cadastrar um insumo novo (ou reforçar a quantidade de um que já existe).
