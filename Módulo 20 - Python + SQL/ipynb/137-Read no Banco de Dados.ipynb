## 🔌 1. Conectando

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
cursor = conexao.cursor()
```

## 📋 2. SELECT * — Trazendo Tudo

`SELECT * FROM tabela` traz todas as colunas de todas as linhas. Cada linha devolvida por `fetchall()` é um objeto `pyodbc.Row`, acessável por índice (`linha[0]`) ou pelo nome da coluna (`linha.Produto`).

```python
cursor.execute("SELECT * FROM dbo.Vendas")
todas_as_vendas = cursor.fetchall()

print(f"{CinzaClaro}Total de linhas:{Reset} {MagentaClaro}{len(todas_as_vendas)}{Reset}\n")
for venda in todas_as_vendas:
    print(f"{CinzaClaro}#{venda.Id}{Reset} {VerdeClaro}{venda.Produto}{Reset} — {venda.Quantidade}x R${venda.ValorUnitario} ({venda.Vendedor})")
```

## 🔁 3. fetchone() vs. fetchall() vs. Iterar o Cursor

| Forma 🔑 | Comportamento 🔓 |
|---|---|
| `cursor.fetchall()` | traz tudo de uma vez, numa lista — ocupa memória proporcional ao tamanho do resultado |
| `cursor.fetchone()` | traz **uma linha por chamada** — útil quando o resultado é gigante e não cabe todo na memória |
| `for linha in cursor:` | itera direto no cursor, uma linha por vez, sem guardar tudo numa lista |

```python
cursor.execute("SELECT Produto, ValorUnitario FROM dbo.Vendas")

primeira_linha = cursor.fetchone()
print(f"{CinzaClaro}Primeira linha (fetchone):{Reset} {VerdeClaro}{primeira_linha.Produto}{Reset}\n")

print(f"{CinzaClaro}Restante, iterando direto no cursor:{Reset}")
for linha in cursor:
    print(f"  {VerdeClaro}{linha.Produto}{Reset} — R${linha.ValorUnitario}")
```

## 🔍 4. Filtrando com WHERE

Igual em qualquer `SELECT`, o `WHERE` filtra as linhas — e os valores continuam entrando como parâmetro (`?`), nunca concatenados na string.

| Operador 🔑 | Exemplo 🔓 |
|---|---|
| `=` | `WHERE Categoria = ?` |
| `>` / `<` | `WHERE ValorUnitario > ?` |
| `LIKE` | `WHERE Produto LIKE ?` (com `%` como coringa) |

```python
cursor.execute(
    "SELECT Produto, Categoria, ValorUnitario FROM dbo.Vendas WHERE Categoria = ?",
    "Eletrônicos"
)
eletronicos = cursor.fetchall()

print(f"{CinzaClaro}Vendas de Eletrônicos:{Reset} {MagentaClaro}{len(eletronicos)}{Reset}")
for venda in eletronicos:
    print(f"  {VerdeClaro}{venda.Produto}{Reset} — R${venda.ValorUnitario}")
```

## 🔢 5. Ordenando com ORDER BY

`ORDER BY coluna DESC` traz do maior pro menor (`ASC`, o padrão, faz o contrário).

```python
cursor.execute("SELECT Produto, ValorUnitario FROM dbo.Vendas ORDER BY ValorUnitario DESC")
vendas_ordenadas = cursor.fetchall()

print(f"{CinzaClaro}Vendas da mais cara pra mais barata:{Reset}")
for venda in vendas_ordenadas:
    print(f"  {VerdeClaro}{venda.Produto}{Reset} — R${venda.ValorUnitario}")

conexao.close()
```

Ler com `cursor` funciona bem — mas quando o objetivo é **analisar** os dados (agrupar, somar, cruzar colunas), o pandas facilita muito mais do que percorrer `pyodbc.Row` na mão. É o assunto do próximo notebook.
