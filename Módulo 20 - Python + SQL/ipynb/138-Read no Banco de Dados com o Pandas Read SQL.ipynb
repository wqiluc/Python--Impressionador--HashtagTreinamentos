## 🔌 1. Conectando

A conexão continua sendo `pyodbc` — o pandas só usa ela por baixo, não substitui o driver.

```python
import pandas as pd
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
```

## 🐼 2. pd.read_sql — A Tabela Inteira

| Parâmetro 🔑 | Papel 🔓 |
|---|---|
| `sql` | o comando `SELECT` (ou nome de uma tabela) |
| `con` | a conexão já aberta |

O resultado já vem com nomes de coluna, tipos inferidos e índice — pronto pra qualquer método de `DataFrame`.

```python
df_vendas = pd.read_sql("SELECT * FROM dbo.Vendas", conexao)

print(f"{CinzaClaro}Formato do DataFrame:{Reset} {MagentaClaro}{df_vendas.shape}{Reset}")
df_vendas
```

## 🎯 3. pd.read_sql com Filtro Parametrizado

`params` recebe os valores que substituem os `?` do SQL — mesma lógica segura de parâmetro do `pyodbc`, só que dentro do pandas.

```python
df_moveis = pd.read_sql(
    "SELECT Produto, Quantidade, ValorUnitario FROM dbo.Vendas WHERE Categoria = ?",
    conexao,
    params=["Móveis"]
)

df_moveis
```

## 📊 4. Analisando o DataFrame Resultante

Com os dados em `DataFrame`, entram todos os métodos de pandas já vistos nos módulos anteriores — sem precisar escrever SQL mais complexo pra isso.

```python
df_vendas["Total"] = df_vendas["Quantidade"] * df_vendas["ValorUnitario"]

faturamento_por_categoria = df_vendas.groupby("Categoria")["Total"].sum().sort_values(ascending=False)

print(f"{CinzaClaro}Faturamento por categoria:{Reset}")
for categoria, total in faturamento_por_categoria.items():
    print(f"  {VerdeClaro}{categoria}{Reset}: R$ {total:,.2f}")

conexao.close()
```

Sempre que o objetivo for **ler pra analisar**, `pd.read_sql` economiza o trabalho de montar o `DataFrame` na mão. Os próximos dois notebooks fecham o CRUD básico com Update e Delete.
