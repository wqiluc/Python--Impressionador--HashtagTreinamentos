## 🔌 1. Conectando

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL", autocommit=True)
cursor = conexao.cursor()
```

## 🔍 2. Construindo buscar_insumo()

Consulta pontual: dado um nome, devolve os dados do insumo (ou `None`, se não encontrar).

```python
def buscar_insumo(nome):
    cursor.execute(
        "SELECT Insumo, Quantidade, UnidadeMedida, Categoria, AtualizadoEm FROM dbo.Estoque WHERE Insumo = ?",
        nome
    )
    resultado = cursor.fetchone()

    if not resultado:
        print(f"{VermelhoClaro}'{nome}' não encontrado no estoque.{Reset}")
        return None

    print(f"{VerdeClaro}{resultado.Insumo}{Reset} — {resultado.Quantidade} {resultado.UnidadeMedida} {CinzaEscuro}({resultado.Categoria}, atualizado em {resultado.AtualizadoEm}){Reset}")
    return resultado
```

## 📋 3. Construindo listar_estoque() com Alerta de Estoque Baixo

`listar_estoque()` traz tudo, ordenado por categoria, e destaca em vermelho os insumos abaixo de um limite mínimo — útil pra saber o que precisa repor antes de faltar de vez.

| Parâmetro 🔑 | Papel 🔓 |
|---|---|
| `limite_alerta` | quantidade abaixo da qual o insumo é destacado como "acabando" |

```python
def listar_estoque(limite_alerta=10.0):
    cursor.execute("SELECT Insumo, Quantidade, UnidadeMedida, Categoria FROM dbo.Estoque ORDER BY Categoria, Insumo")
    itens = cursor.fetchall()

    print(f"{CinzaClaro}Estoque atual ({len(itens)} insumos):{Reset}\n")
    for item in itens:
        quantidade = float(item.Quantidade)
        if quantidade < limite_alerta:
            print(f"  {VermelhoClaro}⚠ {item.Insumo}{Reset} — {AmareloClaro}{quantidade} {item.UnidadeMedida} (estoque baixo!){Reset}")
        else:
            print(f"  {VerdeClaro}{item.Insumo}{Reset} — {quantidade} {item.UnidadeMedida} {CinzaEscuro}({item.Categoria}){Reset}")

    return itens
```

## 🎬 4. O Sistema Completo em Ação

Com `buscar_insumo()` e `listar_estoque()` prontas, o controle de estoque de insumos está completo — as 5 funções construídas ao longo do exercício, todas rodando sobre a mesma tabela.

```python
buscar_insumo("Leite")
buscar_insumo("Insumo Inexistente")

print()
_ = listar_estoque(limite_alerta=10.0)

conexao.close()
```
