## 🔌 1. Conectando

```python
from conexao import nova_conexao_sqlserver
from cores import *

conexao = nova_conexao_sqlserver(banco="HashtagCursoSQL")
cursor = conexao.cursor()
```

## 🔎 2. Conferindo Antes de Apagar

Seguindo a regra do notebook anterior: sempre um `SELECT` com a mesma condição do `DELETE`, pra ver exatamente o que vai sumir.

```python
cursor.execute("SELECT Id, Produto, Categoria FROM dbo.Vendas WHERE Produto = ?", "Luminária de Mesa")
linha_a_apagar = cursor.fetchone()

print(f"{AmareloClaro}Prestes a apagar:{Reset} #{linha_a_apagar.Id} {VermelhoClaro}{linha_a_apagar.Produto}{Reset} ({linha_a_apagar.Categoria})")
```

## 🗑️ 3. DELETE de um Registro

Filtrando pelo `Id`, o `DELETE` remove exatamente a linha conferida no passo anterior.

```python
cursor.execute("DELETE FROM dbo.Vendas WHERE Id = ?", linha_a_apagar.Id)
conexao.commit()

print(f"{VermelhoClaro}{cursor.rowcount} linha apagada.{Reset}")
```

## 📦 4. DELETE com Condição Mais Ampla

Assim como no `UPDATE`, o `WHERE` de um `DELETE` pode atingir várias linhas de uma vez — por isso o `SELECT COUNT(*)` de conferência é ainda mais importante aqui.

```python
cursor.execute("SELECT COUNT(*) AS Total FROM dbo.Vendas WHERE Quantidade < ?", 2)
quantas_seriam_apagadas = cursor.fetchone().Total
print(f"{CinzaClaro}Vendas com menos de 2 unidades:{Reset} {AmareloClaro}{quantas_seriam_apagadas}{Reset}")

cursor.execute("DELETE FROM dbo.Vendas WHERE Quantidade < ?", 2)
conexao.commit()

print(f"{VermelhoClaro}{cursor.rowcount} linha(s) apagada(s).{Reset}")
```

## 🧾 5. Resumo do CRUD com pyodbc

| Operação 🔑 | Comando SQL 🔓 | Método pyodbc 🔓 |
|---|---|---|
| Create | `INSERT INTO` | `execute()` / `executemany()` + `commit()` |
| Read | `SELECT` | `execute()` + `fetchall()` / `fetchone()` |
| Update | `UPDATE ... SET ... WHERE` | `execute()` + `commit()` |
| Delete | `DELETE FROM ... WHERE` | `execute()` + `commit()` |

Todo `INSERT`/`UPDATE`/`DELETE` segue o mesmo fechamento: `execute()` com parâmetros seguros, depois `commit()` pra confirmar.

```python
cursor.execute("SELECT COUNT(*) AS Total FROM dbo.Vendas")
total_final = cursor.fetchone().Total

print(f"{CinzaClaro}Linhas restantes em Vendas:{Reset} {MagentaClaro}{total_final}{Reset}")

conexao.close()
```
