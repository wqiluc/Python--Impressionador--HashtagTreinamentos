## 📥 1. Importando o pathlib

Tudo começa importando a classe `Path` do módulo `pathlib` — é ela quem representa um caminho (de pasta ou arquivo) e concentra os métodos que vamos usar no resto do notebook.

## 📍 2. Descobrindo a Pasta Atual

Antes de navegar pra qualquer lugar, é útil saber onde o próprio script está sendo executado — `Path.cwd()` devolve esse caminho.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `Path.cwd()` | devolve o caminho da pasta em que o Python está rodando no momento |

## 🧭 3. Navegando até uma Pasta Específica

Pra apontar pra uma pasta (ou arquivo) que não é a atual, basta passar o caminho como texto pro `Path()` — usando `/` como separador, mesmo em notebooks rodando no Windows.

| Forma 🔑 | Resultado 🔓 |
|---|---|
| `Path('pasta1/pasta2/pasta3')` | caminho relativo, a partir da pasta atual |
| `Path('C:/pasta1/pasta2')` | caminho absoluto, a partir da raiz do sistema |

> ⚠️ **Atenção:** Windows normalmente usa `\` e Mac/Linux usa `/` — o `pathlib` resolve essa diferença sozinho, então usar `/` no código funciona nos três sistemas.

## 📄 4. Listando os Arquivos de uma Pasta

Com o caminho de uma pasta em mãos, `iterdir()` percorre tudo o que existe dentro dela — arquivos e subpastas.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `pasta.iterdir()` | devolve um iterável com cada item (arquivo ou subpasta) dentro da pasta |
| `list(pasta.iterdir())` | transforma esse iterável numa lista de `Path`, pra poder percorrer/ordenar |

> 💡 **Dica:** `iterdir()` não entra nas subpastas sozinho — pra isso existe o `.rglob('*')`, que percorre tudo recursivamente.

## 📁 5. Criando uma Pasta Nova

Pra criar uma pasta nova, chama-se `.mkdir()` diretamente no `Path` que representa o caminho desejado (a pasta ainda não precisa existir).

| Método 🔑 | O que faz 🔓 |
|---|---|
| `pasta.mkdir()` | cria a pasta representada por esse `Path` |
| `pasta.mkdir(exist_ok=True)` | cria a pasta, mas não dá erro se ela já existir |
| `pasta.mkdir(parents=True)` | cria também as pastas intermediárias do caminho que ainda não existem |

> ⚠️ **Atenção:** sem `exist_ok=True`, chamar `.mkdir()` numa pasta que já existe lança `FileExistsError`.

## ✅ 6. Verificando se um Arquivo Existe

Antes de copiar, mover ou até ler um arquivo, vale conferir se ele realmente existe no caminho esperado — evita erro na hora de mexer em algo que não está lá.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `caminho.exists()` | devolve `True`/`False` conforme o arquivo (ou pasta) existe ou não |
| `caminho.is_file()` | devolve `True` só se o caminho existir **e** for um arquivo (não uma pasta) |

## 📋 7. Copiando um Arquivo

Copiar e mover não são responsabilidade do `pathlib` — pra essas ações entra o módulo `shutil`, que faz esse tipo de operação com arquivo com bem menos chance de erro do que remontar tudo na mão.

| Método 🔑 | O que faz 🔓 |
|---|---|
| `shutil.copy2(origem, destino)` | copia o arquivo de `origem` pra `destino`, mantendo metadados (data de criação, permissões) |

> 💡 **Dica:** `origem` e `destino` podem ser tanto texto quanto `Path` — o `shutil` aceita os dois.

## 🔀 8. Movendo um Arquivo

Existem 2 formas de mover um arquivo: renomeando o caminho dele com o próprio `pathlib`, ou usando o `shutil.move()`.

| Forma 🔑 | Como funciona 🔓 |
|---|---|
| `caminho.rename(caminho_novo)` | move o arquivo pra outro caminho — funciona também pra renomear, se o caminho novo estiver na mesma pasta |
| `shutil.move(origem, destino)` | move o arquivo, com mais tolerância que o `rename()` |

> ⚠️ **Atenção:** `.rename()` não cria a pasta de destino sozinho — ela precisa já existir, senão dá erro.

## 🛠️ 9. Aplicando na Prática

Hora de sair do exemplo genérico e aplicar cada método na pasta `spec`, que guarda os indicadores trimestrais das 18 lojas de shopping.

### 📄 Listando os arquivos da pasta `spec`

```python
from pathlib import Path
from cores import *

pasta_spec = Path("../spec")
arquivos_csv = sorted(pasta_spec.glob("*.csv"))

print(f"{CinzaClaro}Arquivos encontrados:{Reset} {MagentaClaro}{len(arquivos_csv)}{Reset}")
for indice, arquivo in enumerate(arquivos_csv[:5]):
    print(f"{CinzaClaro}{Reset}{MagentaClaro}{arquivo.name}{Reset}")
print(f"{CinzaClaro}...{Reset}")
```

### ✅ Verificando se um arquivo existe na pasta

```python
arquivo_esperado = pasta_spec / "201801_Ibirapuera_SP.csv"
arquivo_inexistente = pasta_spec / "201801_Loja_Fantasma_XX.csv"

print(f"{CinzaClaro}{arquivo_esperado.name} existe?{Reset} {VerdeClaro}{arquivo_esperado.exists()}{Reset}")
print(f"{CinzaClaro}{arquivo_inexistente.name} existe?{Reset} {VermelhoClaro}{arquivo_inexistente.exists()}{Reset}")
```

### 📁 Criando uma pasta nova

```python
pasta_backup = pasta_spec / "Backup"
pasta_backup.mkdir(exist_ok=True)

print(f"{CinzaClaro}Pasta criada:{Reset} {MagentaClaro}{pasta_backup}{Reset}")
print(f"{CinzaClaro}Existe agora?{Reset} {VerdeClaro}{pasta_backup.exists()}{Reset}")
```

### 📋 Criando uma cópia do nosso arquivo na nova pasta

```python
import shutil

arquivo_original = pasta_spec / "201801_Ibirapuera_SP.csv"
arquivo_copiado = pasta_backup / arquivo_original.name

shutil.copy2(arquivo_original, arquivo_copiado)

print(f"{CinzaClaro}Cópia criada?{Reset} {VerdeClaro}{arquivo_copiado.exists()}{Reset}")
```

### 🔀 Movendo um arquivo de um lugar para o outro

```python
pasta_arquivados = pasta_backup / "Arquivados"
pasta_arquivados.mkdir(exist_ok=True)

arquivo_movido = arquivo_copiado.rename(pasta_arquivados / arquivo_copiado.name)

print(f"{CinzaClaro}Ainda está em Backup?{Reset} {VermelhoClaro}{arquivo_copiado.exists()}{Reset}")
print(f"{CinzaClaro}Está em Backup/Arquivados?{Reset} {VerdeClaro}{arquivo_movido.exists()}{Reset}")
```

## 🎯 10. Desafio: Separando os Indicadores por Estado

Você trabalha em uma empresa que tem 18 lojas espalhadas pelo Brasil, divididas em 5 estados diferentes:

| Estado 🔑 | Sigla 🔓 |
|---|---|
| Rio de Janeiro | `RJ` |
| São Paulo | `SP` |
| Minas Gerais | `MG` |
| Goiás | `GO` |
| Amazonas | `AM` |

Todo trimestre, são calculados os indicadores de cada funcionário de cada loja, e esses indicadores ficam armazenados em um arquivo — no nosso caso, um `.csv` por loja/mês, todos juntos na pasta `spec`.

Cada estado tem 1 Gerente Geral, responsável por todas as lojas daquele estado. Pediram pra você separar as bases de indicadores de forma que cada Gerente Geral receba só os arquivos das lojas do estado dele, ou seja, organizar tudo em uma pasta por estado.

> 💡 **Dica:** o nome de cada arquivo já traz a sigla do estado no final (ex.: `201801_Ibirapuera_SP.csv`) — dá pra usar `arquivo.stem.split('_')[-1]` pra extrair a sigla, sem precisar abrir o conteúdo do arquivo.

> ⚠️ **Atenção:** ainda não vimos como enviar e-mail pelo Python — por enquanto, o objetivo é só deixar os arquivos já separados, prontos pra cada gerente.

```python
from pathlib import Path
import shutil
from cores import *

pasta_spec = Path("../spec")
pasta_por_estado = pasta_spec / "Por_Estado"

arquivos_csv = sorted(pasta_spec.glob("*.csv"))
contagem_por_estado = {}

for indice_arquivo, arquivo in enumerate(arquivos_csv):
    sigla_estado = arquivo.stem.split("_")[-1]
    pasta_estado = pasta_por_estado / sigla_estado
    pasta_estado.mkdir(parents=True, exist_ok=True)

    shutil.copy2(arquivo, pasta_estado / arquivo.name)
    contagem_por_estado[sigla_estado] = contagem_por_estado.get(sigla_estado, 0) + 1

print(f"{CinzaClaro}Total de arquivos organizados:{Reset} {MagentaClaro}{sum(contagem_por_estado.values())}{Reset}\n")
for estado in sorted(contagem_por_estado):
    print(f"{CinzaClaro}{estado}:{Reset} {VerdeClaro}{contagem_por_estado[estado]} arquivos{Reset}")
```
